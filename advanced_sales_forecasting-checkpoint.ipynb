# Advanced Sales Forecasting with Real Retail Transactions
**Python · Pandas · Feature engineering · Gradient boosting · Streamlit**

Source: UCI Machine Learning Repository, Online Retail. Transactions are converted to daily revenue (`Quantity × UnitPrice`). Cancelled invoices and non-positive quantities/prices are excluded for this gross-sales exercise.

> Forecasts are uncertain and for educational analysis only.

## 1. Install dependencies
From the project root, run `pip install -r requirements.txt`. In Jupyter, use `%pip install pandas numpy matplotlib seaborn scikit-learn openpyxl requests` if needed.

```python
from pathlib import Path
import io, zipfile, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import requests
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.pipeline import make_pipeline

warnings.filterwarnings("ignore", category=FutureWarning)
sns.set_theme()
ROOT = Path.cwd()
if ROOT.name == "notebooks": ROOT = ROOT.parent
DATA_DIR, OUTPUT_DIR = ROOT/"data", ROOT/"outputs"
DATA_DIR.mkdir(exist_ok=True); OUTPUT_DIR.mkdir(exist_ok=True)
print("Project root:", ROOT.resolve())
```

## 2. Download public UCI dataset
Dataset page: https://archive.ics.uci.edu/dataset/352/online+retail. If the automatic download fails, download the Excel workbook from the page and save it as `data/Online Retail.xlsx`.

```python
URL = "https://archive.ics.uci.edu/static/public/352/online+retail.zip"
xlsx = DATA_DIR/"Online Retail.xlsx"
if not xlsx.exists():
    response = requests.get(URL, timeout=60)
    response.raise_for_status()
    with zipfile.ZipFile(io.BytesIO(response.content)) as z:
        names = [n for n in z.namelist() if n.lower().endswith((".xlsx",".xls"))]
        if not names: raise ValueError("No Excel workbook found in the UCI archive.")
        xlsx.write_bytes(z.read(names[0]))
transactions = pd.read_excel(xlsx)
print("Shape:", transactions.shape)
display(transactions.head())
display(transactions.dtypes.to_frame("dtype"))
```

## 3. Clean transactions and aggregate to daily revenue

```python
transactions.columns = [str(c).strip() for c in transactions.columns]
required = {"InvoiceDate","Quantity","UnitPrice"}
if not required.issubset(transactions.columns):
    raise ValueError(f"Missing columns: {required-set(transactions.columns)}")
transactions["InvoiceDate"] = pd.to_datetime(transactions["InvoiceDate"], errors="coerce")
transactions["Quantity"] = pd.to_numeric(transactions["Quantity"], errors="coerce")
transactions["UnitPrice"] = pd.to_numeric(transactions["UnitPrice"], errors="coerce")
clean = transactions.dropna(subset=["InvoiceDate","Quantity","UnitPrice"]).copy()
if "InvoiceNo" in clean:
    clean = clean[~clean["InvoiceNo"].astype(str).str.startswith("C", na=False)]
clean = clean[(clean["Quantity"]>0)&(clean["UnitPrice"]>0)]
clean["Revenue"] = clean["Quantity"]*clean["UnitPrice"]
daily = clean.set_index("InvoiceDate")["Revenue"].resample("D").sum().asfreq("D").fillna(0).to_frame("Sales")
daily.index.name = "Date"
print("Original rows:",len(transactions),"Clean rows:",len(clean),"Daily rows:",len(daily))
print("Period:",daily.index.min().date(),"to",daily.index.max().date())
display(daily.head())
```

## 4. Exploratory data analysis

```python
fig, axes = plt.subplots(2,1,figsize=(14,8),sharex=True)
axes[0].plot(daily.index,daily.Sales,label="Daily revenue")
axes[0].plot(daily.index,daily.Sales.rolling(7).mean(),label="7-day rolling mean")
axes[0].set_title("Daily revenue trend"); axes[0].set_ylabel("Revenue"); axes[0].legend()
axes[1].bar(daily.index,daily.Sales,width=1); axes[1].set_title("Daily revenue"); axes[1].set_ylabel("Revenue")
plt.tight_layout(); plt.show()
weekday = daily.groupby(daily.index.day_name()).Sales.mean().reindex(["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"])
weekday.plot(kind="bar",figsize=(10,4),title="Average revenue by weekday"); plt.ylabel("Revenue"); plt.tight_layout(); plt.show()
display(daily.Sales.describe().to_frame("Daily revenue statistics"))
```

## 5. Feature engineering
Lag and rolling features use only prior days (rolling features are shifted by one day) to avoid target leakage.

```python
def make_features(series):
    f = pd.DataFrame({"Sales":series.astype(float)})
    ix=f.index
    f["dow"]=ix.dayofweek; f["day"]=ix.day; f["month"]=ix.month
    f["quarter"]=ix.quarter; f["year"]=ix.year; f["dayofyear"]=ix.dayofyear
    f["weekend"]=(ix.dayofweek>=5).astype(int)
    for lag in [1,2,3,7,14,21,28]: f[f"lag_{lag}"]=f.Sales.shift(lag)
    for w in [7,14,28]:
        f[f"rollmean_{w}"]=f.Sales.shift(1).rolling(w).mean()
        f[f"rollstd_{w}"]=f.Sales.shift(1).rolling(w).std()
    return f.replace([np.inf,-np.inf],np.nan)

feat = make_features(daily.Sales)
feature_cols = [c for c in feat.columns if c!="Sales"]
data = feat.dropna(subset=feature_cols+["Sales"])
print("Usable rows:",len(data),"Feature count:",len(feature_cols))
display(data.head())
```

## 6. Chronological holdout evaluation
First 80% of dates train the model; the final 20% are holdout data. The baseline predicts each day using the sales from seven calendar days earlier.

```python
if len(data)<100: raise ValueError("Need at least 100 usable daily observations.")
split=int(len(data)*.8); train,test=data.iloc[:split],data.iloc[split:]
model=make_pipeline(SimpleImputer(strategy="median"),
    HistGradientBoostingRegressor(max_iter=250,learning_rate=.06,max_leaf_nodes=15,l2_regularization=1.0,random_state=42))
model.fit(train[feature_cols],train.Sales)
pred=np.maximum(0,model.predict(test[feature_cols]))
baseline=test["lag_7"].to_numpy()
actual=test.Sales.to_numpy()
def metrics(y,p):
    return {"MAE":mean_absolute_error(y,p),"RMSE":np.sqrt(mean_squared_error(y,p)),
            "WAPE_%":np.abs(y-p).sum()/max(np.abs(y).sum(),1e-9)*100}
scores=pd.DataFrame({"HistGradientBoosting":metrics(actual,pred),"Weekly seasonal naive":metrics(actual,baseline)}).T
display(scores.round(3))
results=pd.DataFrame({"Actual":actual,"Model":pred,"Weekly_naive":baseline},index=test.index)
display(results.head())
```

```python
results.plot(figsize=(14,5),title="Holdout: actual vs forecast")
plt.ylabel("Revenue"); plt.tight_layout(); plt.show()
print("Model MAE:",round(scores.loc["HistGradientBoosting","MAE"],3))
print("Baseline MAE:",round(scores.loc["Weekly seasonal naive","MAE"],3))
print("Model lower MAE than baseline:",bool(scores.loc["HistGradientBoosting","MAE"]<scores.loc["Weekly seasonal naive","MAE"]))
```

## 7. Recursive forecast for the next 30 calendar days
Each forecast becomes a lagged input for later forecasts, so errors can compound. Promotions, holidays, pricing, and external drivers are not included.

```python
horizon=30
history=daily.Sales.astype(float).copy()
future_dates=pd.date_range(history.index[-1]+pd.Timedelta(days=1),periods=horizon,freq="D")
values=[]
for dt in future_dates:
    extended=pd.concat([history,pd.Series(index=[dt],dtype=float)])
    row=make_features(extended).loc[[dt],feature_cols]
    value=max(0.0,float(model.predict(row)[0]))
    history.loc[dt]=value; values.append(value)
future=pd.DataFrame({"Forecast":values},index=future_dates); future.index.name="Date"
display(future.head(10))
plt.figure(figsize=(14,5))
plt.plot(daily.Sales.tail(90),label="Recent actual")
plt.plot(future.Forecast,label="Future forecast",linestyle="--")
plt.legend(); plt.ylabel("Revenue"); plt.tight_layout(); plt.show()
```

## 8. Export results

```python
daily.to_csv(DATA_DIR/"daily_sales.csv")
results.to_csv(OUTPUT_DIR/"holdout_predictions.csv")
future.to_csv(OUTPUT_DIR/"future_sales_forecast.csv")
scores.to_csv(OUTPUT_DIR/"model_metrics.csv")
print("Saved data and outputs.")
```
